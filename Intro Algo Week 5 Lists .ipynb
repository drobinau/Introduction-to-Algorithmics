```python
import random 

n = 10000
count = [0]*6

for i in range(n):
    die1 = random.randint(1,6)
    die2 = random.randint(1,6)
    diff = abs(die1-die2)
    count[diff] += 1 

print('The number of each difference is: ', count)
print('The most likely difference is: ', count.index(max(count)))
```

```python
import random

n = 1000
rollCount = []

for i in range(n):
    count = 0
    seenFaces = []

    while len(seenFaces) < 6:
        roll = random.randint(1, 6)

        if roll not in seenFaces:
            seenFaces.append(roll)

        count += 1

    rollCount.append(count)

average_rolls = sum(rollCount) / n

print("The average number of rolls is:", average_rolls)
```
